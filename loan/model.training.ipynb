```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,confusion_matrix, classification_report,precision_score,recall_score,f1_score
import joblib
```

```python
df=pd.read_csv('D:\ml project2\loan_approval_dataset.csv')
df
```

```python
x = df.drop("Loan_Approved", axis=1)
y = df["Loan_Approved"]
```

```python
num_cols = ["Age", "Salary","Credit_Score","Loan_Amount"]
cat_cols = ["Loan_Term","Employment_Status","Residence_Type","Previous_Default"]
```

```python
df.info()
```

```python
num_pipeline = Pipeline(steps=[("imputer", SimpleImputer(strategy="median"))])
```

```python
cat_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_pipeline, num_cols),
        ("cat", cat_pipeline, cat_cols)
    ]
)
```

```python
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=200,
            random_state=42
        ))
    ]
)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
model.fit(x_train, y_train)
```

```python
y_pred_train=model.predict(x_train)
```

```python
y_pred_test = model.predict(x_test)
```

```python
accuracy_score(y_pred_train,y_train)
```

```python
accuracy_score(y_pred_test,y_test)
```

```python
precision_score(y_test, y_pred_test, pos_label="Yes")
```

```python
recall_score(y_test, y_pred_test, pos_label="Yes")
```

```python
f1_score(y_test, y_pred_test, pos_label="Yes")
```

```python
confusion_matrix(y_test, y_pred_test)
```

```python
classification_report(y_test, y_pred_test)
```

```python
joblib.dump(model,'model_trained.pk3')
print('Model Trained and Saved')
```
